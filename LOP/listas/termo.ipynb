```python
import random

palavras = [
    "Horizonte",
    "lampada",
    "tempestade",
    "abacaxi",
    "labirinto",
    "caderno",
    "neblina",
    "foguete",
    "sussurro",
    "esmeralda"
]
palavra = random.choice(palavras).lower()
listaLetras = []
tentativa = 1

for letra in palavra:
    listaLetras.append(letra)

print(f"Palavra de {len(listaLetras)} letras")

while True:
    if tentativa > 5:
        print('\nSuas tentetivas acabaram tente novamente !!!')
        break
    else:
        print(f'\nTentativa numero {tentativa}\n')

    palavraTentativa = input("Digite uma palavra: ")
    palavraTentativa = palavraTentativa.lower()
    listaTentativa = []

    print(f'Sua tentativa foi: {palavraTentativa}')

    for letraTentativa in palavraTentativa:
        listaTentativa.append(letraTentativa)

    posicaoCerta = 0
    if len(listaTentativa) == len(listaLetras):
        for i in range(len(listaTentativa)):
            if listaTentativa[i] == listaLetras[i]:
                posicaoCerta += 1
                print(f'A letra "{listaTentativa[i]}" na posicao {i + 1} esta correta')
    else:
        print(f"Palavra de {len(listaLetras)} letras")
        tentativa += 1
        continue
    
    
    if posicaoCerta != len(listaLetras):
        print(f'{posicaoCerta}/{len(listaLetras)} de letras na posicao certa')
    else:
        print(f'\nParabens voce acertou a palavra era {palavra}\n')
        break

    tentativa += 1
```
