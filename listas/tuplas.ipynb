```python
vendas_brutas = [
    (101, "Monitor", 5, 1200.0),
    (102, "Mouse", 50, 80.0),
    (103, "Teclado", 8, 150.0),
    (104, "Case", 3, 50.0),
]
totalInven = 0
reajuste = 1.10
nova_vendas_brutas = []

print("Filtrar produtos com estoque baixo: ")
for cada in vendas_brutas:
    if cada[2] < 10:
        print(cada)

print("\nCalcular o valor total do inventário: ")
for total in vendas_brutas:
    valor = total[3] * total[2]
    totalInven += valor
print(f"R$ {totalInven}")

print("\nLista de tuplas atualizada com um reajuste de preço.: ")
for cada in vendas_brutas:
    novoPreco = cada[3] * reajuste
    nova_vendas_brutas.append((cada[0], cada[1], cada[2], novoPreco))
print(nova_vendas_brutas)
```
