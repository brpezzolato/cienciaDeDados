```python
nomeFilial = input("Digite o nome da filial: ")
listaVenda = []
listaVendaLimpa = []

while True:
    valorVenda = int(input("Digite um valor de venda: "))
    if valorVenda == 0:
        break
    else:
        listaVenda.append(valorVenda)


def limpar_vendas(listaVenda):
    for cada in listaVenda:
        if cada < 0 or cada > 10000:
            continue
        else:
            listaVendaLimpa.append(cada)


limpar_vendas(listaVenda)


def analisar_dados(listaLimpa):
    soma = sum(listaLimpa)
    media = sum(listaLimpa) / len(listaLimpa)

    return (soma, media)


resposta = analisar_dados(listaVendaLimpa)


def definir_status(media):
    if media > 5000:
        return "Alta Performance"
    elif media >= 2000 and media <= 5000:
        return "Performance Estável"
    else:
        return "Performance Crítica"

status = definir_status(resposta[1])

print('\nResumo das vendas\n')

print(f'Nome franquia: {nomeFilial}')
print(f'O status da sua franquia e: {status}')
print(f'Faturamento total: {resposta[0]:.2f}')
print(f'Media de vendas: {resposta[1]:.2f}')

print('\nLista de vendas ordenadas')
count = 1
for i in listaVendaLimpa:
    print(f'Venda numero {count}: {i}')
    count += 1
```
