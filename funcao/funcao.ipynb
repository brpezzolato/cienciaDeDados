Função para pegar o maior numero dentro de 5

```python
def maior():
    maiorNum = 0
    for i in range(5):
        num = float(input('Digite um numero: '))
        if i == 0:
            antigo = num
        if num > antigo:
            maiorNum = num
    print(maiorNum)


maior()
```

Função para calcular o fatorial de um numero

```python
def fatorial():
    num = int(input("Digite um numero e descubra o fatorial: "))
    fatorial = 1

    for i in range(1, num + 1):
        fatorial *= i

    print(fatorial)


fatorial()
```

Função para ver quantas vogais ha em uma string

```python
vogais = "aeiou"
palavra = input("Digite uma palavra: ").lower()

def Vogal(palavra):
    count = 0
    for i in palavra:
        if i in vogais:
            count += 1
    return count

print(Vogal(palavra))
```

Função para calcular raiz quadrada

```python
num = int(input("Digite um numero para saber a raiz quadrada: "))


def raizQuadrada(num):
    return num**0.5


print(f'A raiz quadrada de {num} e = {raizQuadrada(num)}')
```

Valores Default

```python
def calc_bonus (venda, taxa = 1000, percentual = 0.05):
    if venda >= taxa:
        return venda * percentual
    else:
        return 0.0
    
vendaAnalista = 1200
bonus = calc_bonus(vendaAnalista)
print(f"O bonus calculado para R$ {vendaAnalista} foi R$ {bonus}")
```

Funçao com lista

```python
def overviewLista(lista):
    minimo = min(lista)
    maximo = max(lista)
    media = sum(lista) / len(lista)
    return minimo, maximo, media


vendas_mes = [1200, 1500, 1100, 1900]

res = overviewLista(vendas_mes) #RETORNA UMA TUPLA
minimo, maximo, media = overviewLista(vendas_mes) #ARMAZENA CADA VALOR DE ACORDO COM A ORDEM DO RETURN

print(f'Pior valor: {res[0] or minimo}')
print(f'Maior valor: {res[1] or maximo}')
print(f'Media: {res[2] or media}')
```
